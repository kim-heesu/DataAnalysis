```python
import requests

url ='https://apis.data.go.kr/B552845/perDay/price'
api_key ='4825272ae11d0'

vrty_value= ['01','02','03','06']

for value in vrty_value:
    params = {
        'serviceKey' : api_key,
        'numOfRows' : '200',
        'returnType' : 'JSON',
        'cond[exmn_ymd::GTE]' :'20150101',
        'cond[exmn_ymd::LTE]' : '20241231',
        'cond[se_cd::EQ]' :'02',
        'cond[ctgry_cd::EQ]' : '200',
        'cond[item_cd::EQ]' : '211',
        'cond[grd_cd::EQ]' : '04',
        'cond[sgg_cd::EQ]' : '1101',
        'cond[mrkt_cd::EQ]' : '0110211',
        'cond[vrty_cd::EQ]' : value
    }
    response = requests.get(url, params=params)
    print(response.json())


```
