연도별, 계절별 분석을 위해 조사일자(exmn_ymd) 컬럼을 활용하여 연도(year)와 계절
(season) 컬럼을 추가하는 전처리 코드를 작성하고, 변환 결과를 확인할 수 있는 출력 결과를 첨부하시오. (4점)
※ 계절 구분: 봄(3-5월), 여름(6-8월), 가을(9-11월), 겨울(12-2월)

```python
import requests
import pandas as pd

url ='https://apis.data.go.kr/B552845/perDay/price'
api_key =''

params = {
    'serviceKey' : api_key,
    'numOfRows' : '200',
    'returnType' : 'JSON',
    'cond[exmn_ymd::GTE]' :'20150101',
    'cond[exmn_ymd::LTE]' : '20241231',
    'cond[se_cd::EQ]' :'02',
    'cond[ctgry_cd::EQ]' : '200',
    'cond[item_cd::EQ]' : '211',
    'cond[grd_cd::EQ]' : '04',
    'cond[sgg_cd::EQ]' : '1101',
    'cond[mrkt_cd::EQ]' : '0110211',
}
response = requests.get(url, params=params)
result = response.json()

items = result['response']['body']['items']['item']
data.extend(items)

df = pd.DataFrame(data)
df['year'] = [int(date[:4]) for date in df['exmn_ymd']]
month = [int(date[4:6]) for date in df['exmn_ymd']]
season = []

for m in month:
    if 3 <= m <= 5 :
        season.append('봄')
    elif 6 <= m <= 8 :
        season.append('여름')
    elif 9 <= m <= 11 :
        season.append('가을')
    else :
        season.append('겨울')

df['season'] = season
df
```
