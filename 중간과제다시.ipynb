```python
import requests

url='https://apis.data.go.kr/B552845/perDay/price'
api_key = ''

years = range(2015,2025)
vrty_value = ['01','02','03','06']

data = []

for year in years:
    for value in vrty_value:
        params= {
            'serviceKey' : api_key,
            'pageNo' : 1,
            'numOfRows' : '200',
            'returnType' : 'JSON',
            'cond[exmn_ymd::GTE]' :f'{year}0101',
            'cond[exmn_ymd::LTE]' : f'{year}1231',
            'cond[se_cd::EQ]' :'02',
            'cond[ctgry_cd::EQ]' : '200',
            'cond[item_cd::EQ]' : '211',
            'cond[grd_cd::EQ]' : '04',
            'cond[sgg_cd::EQ]' : '1101',
            'cond[mrkt_cd::EQ]' : '0110211',
            'cond[vrty_cd::EQ]' : value
        }
        response = requests.get(url, params=params)
        result =response.json()

        items = result['response']['body']['items']['item']

        data.extend(items)

print(data)
print(len(data))
```

```python
import requests

url = 'https://apis.data.go.kr/B552845/perDay/price'
api_key = ''

params = {
    'serviceKey': api_key,
    'pageNo': 1,
    'numOfRows': '200',
    'returnType': 'JSON',
    'cond[exmn_ymd::GTE]': '20150101',
    'cond[exmn_ymd::LTE]': '20151231',
    'cond[se_cd::EQ]': '02',
    'cond[ctgry_cd::EQ]': '200',
    'cond[item_cd::EQ]': '211',
    'cond[grd_cd::EQ]': '04',
    'cond[sgg_cd::EQ]': '1101',
    'cond[mrkt_cd::EQ]': '0110211',
    'cond[vrty_cd::EQ]': '01'
}

response = requests.get(url, params=params)
result = response.json()

print('API 응답 결과:')
print(result)

print('전체 데이터 건수:')
print(result['response']['body']['totalCount'])
```

```python
import pandas as pd

df = pd.DataFrame(data)

df.info()
```

```python
df['year'] = [int(date[:4]) for date in df['exmn_ymd']]
month = [int(date[4:6]) for date in df['exmn_ymd']]
season=[]
for m in month:
    if 3 <= m <= 5 :
        season.append('봄')
    elif 6 <= m <= 8 :
        season.append('여름')
    elif 9 <= m <= 11 :
        season.append('가을')
    else :
        season.append('겨울')
df['season'] = season
df
```

```python
import matplotlib.pyplot as plt
import pandas as pd

df['exmn_dd_prc'] = pd.to_numeric(df['exmn_dd_prc'])
yearPrice = df.groupby('year')['exmn_dd_prc'].mean()

plt.plot(yearPrice.index, yearPrice.values, marker='o')
plt.title('연도별 배추 평균 가격 변화 - 2760')
plt.xlabel('연도')
plt.ylabel('평균가격')
plt.grid(True)
plt.savefig('연도별배추평균가격변화-2760.png')
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

seasonPrice = df.groupby('season')['exmn_dd_prc'].mean().reset_index()
seasonOrder = ['봄','여름','가을','겨울']

ax = sns.barplot(x='season',y='exmn_dd_prc',data=seasonPrice,order=seasonOrder)

plt.title('계절별 배추 평균 가격')
plt.xlabel('계절')
plt.ylabel('평균가격')

for container in ax.containers:
    ax.bar_label(container, fmt='%.0f')
plt.show()
```
